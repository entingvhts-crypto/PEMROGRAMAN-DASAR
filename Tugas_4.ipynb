```python
#Latihan 4.1

tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

print("Apakah tahun", tahun, "merupakan tahun kabisat?", kabisat)
```

```python
#Latihan 4.2

def hitung_nilai_akhir():

  nilai_tugas = float(input("masukkan nilai tugas: "))
  nilai_uts = float(input("masukkan nilai uts: "))
  nilai_uas = float(input("masukkan nilai uas: "))

  nilai_akhir = (nilai_tugas * 0.30) + (nilai_uts * 0.30) + (nilai_uas * 0.40)

  print("\n--- Hasil Perhitungan ---")
  print("nilai tugas :", nilai_tugas)
  print("nilai uts :", nilai_uts)
  print("nilai uas :", nilai_uas)
  print("nilai akhir :", nilai_akhir)

hitung_nilai_akhir()

```
