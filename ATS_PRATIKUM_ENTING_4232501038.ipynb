```python
#SOAL NO 5 (Katup Pengaman)

tekanan_uap = float(input("Masukkan tekanan uap (psi): "))

# Nilai status manual override
status_manual_override = False

# Menentukan status katup
if status_manual_override:
    print("Katup Mode Manual")
elif tekanan_uap > 150 and not status_manual_override:
    print("Katup Terbuka")
else:
    print("Katup Tertutup")
```

```python
#SOAL NO 7 (Pemanas Mencapai Target)
#Suhu awal
suhu_saat_ini = 60

#Perulangan hingga mencapai target
while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} °C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

# Pesan setelah target tercapai
print(f"Target suhu {suhu_saat_ini} °C tercapai.")
```

PENJELASAN CARA KERJA

Program ini digunakan untuk mensimulasikan proses pemanasan suhu dari 60°C hingga mencapai 85°C. ketika suhu masih di bawah target, program akan terus menaikkan suhu sebesar 5°C menggunakan perulangan while. Setiap kenaikan suhu akan ditampilkan ke layar. Ketika suhu mencapai 80°C, program akan memberikan peringatan bahwa suhu sudah mendekati target. Setelah suhu mencapai 85°C, perulangan berhenti dan program menampilkan pesan bahwa target suhu telah tercapai.

```python
#SOAL NO 9 (Pembacaan Vibrasi dengan Sensor Rusak)

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Variabel untuk menghitung jumlah data tidak valid
data_tidak_valid = 0

# Memeriksa setiap data dengan perulangan for
for unit, nilai in enumerate(data, start=1):
    try:
        # Mengubah data menjadi float
        vibrasi = float(nilai)

        # Memeriksa apakah vibrasi tinggi
        if vibrasi > 2.0:
            print(f"Warning: Unit {unit} vibrasi tinggi!")

        # Memeriksa kondisi kritis
        if vibrasi > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

    except ValueError:
        print(f"Unit {unit}: data tidak valid")
        data_tidak_valid += 1
        continue

# Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {data_tidak_valid}")
```

PENJELASAN CARA KERJA PROGRAM DAN FUNGSI CODE

Program menggunakan list data untuk menyimpan data vibrasi dari lima unit mesin. Perulangan for digunakan untuk memeriksa setiap data secara berurutan dengan nomor unit dimulai dari 1 menggunakan enumerate(). Di dalam blok try-except, program mencoba mengubah data menjadi tipe float. Jika data tidak dapat diubah, seperti "ERR", program akan masuk ke bagian except, menampilkan pesan data tidak valid, dan menambah nilai variabel data_tidak_valid. Selanjutnya, percabangan if digunakan untuk mendeteksi vibrasi tinggi jika nilainya lebih dari 2.0. Jika nilai vibrasi melebihi 3.0, program menampilkan pesan kritis lalu menghentikan pemeriksaan menggunakan break. Setelah perulangan selesai, program menampilkan jumlah data yang tidak valid.

```python
#CHALLENGE
# Program pemantauan konsumsi listrik harian.
# Data diperoleh dari sensor, tetapi ada kemungkinan
# beberapa data tidak valid. Program akan memeriksa
# setiap data, mengelompokkan tingkat konsumsi listrik
# menghitung jumlah data yang tidak valid.

data_konsumsi = ["120", "180", "ERROR", "250", "90"]

data_tidak_valid = 0

for hari, data in enumerate(data_konsumsi, start=1):
    try:
        konsumsi = float(data)

        print(f"Hari ke-{hari}: {konsumsi} kWh")

        if konsumsi > 200:
            print("Status: Konsumsi Tinggi")
        elif konsumsi >= 100:
            print("Status: Konsumsi Normal")
        else:
            print("Status: Konsumsi Rendah")

    except ValueError:
        print(f"Hari ke-{hari}: data tidak valid")
        data_tidak_valid += 1

print(f"\nJumlah data tidak valid: {data_tidak_valid}")
```

PENEJELASAN CARA KERJA PROGRAM

Program ini digunakan untuk memantau konsumsi listrik harian berdasarkan data yang tersimpan dalam sebuah list. Perulangan for digunakan untuk membaca setiap data secara berurutan. Data kemudian ubah ke tipe float menggunakan try-except. Jika data tidak dapat dikonversi, misalnya berisi "ERROR", program akan menampilkan pesan bahwa data tidak valid dan menambah penghitung data tidak valid. Selanjutnya, percabangan if-elif-else digunakan untuk mengelompokkan konsumsi listrik menjadi kategori rendah, normal, atau tinggi berdasarkan nilai konsumsi yang diperoleh. Setelah seluruh data selesai diproses, program menampilkan jumlah data yang tidak valid.
PENJELASAN FUNGSI CODE
- List (data_konsumsi) digunakan untuk menyimpan beberapa data konsumsi listrik harian dalam satu variabel sehingga data lebih mudah dikelola dan diproses secara berurutan.
- Perulangan for digunakan untuk membaca setiap data yang ada di dalam list satu per satu tanpa perlu menuliskan kode yang sama berulang kali. Variabel hari digunakan untuk menunjukkan urutan hari, sedangkan data digunakan untuk mengambil nilai pada setiap iterasi.
- Blok try-except digunakan untuk menangani kemungkinan adanya data yang tidak valid. Pada bagian try, program mencoba mengubah data menjadi tipe float agar dapat dilakukan perhitungan. Jika data tidak bisa diubah menjadi angka, misalnya berisi "ERROR", maka program akan masuk ke bagian except dan menampilkan pesan bahwa data tersebut tidak valid tanpa menghentikan program.
- Percabangan if-elif-else digunakan untuk mengelompokkan tingkat konsumsi listrik berdasarkan nilai yang diperoleh. Jika konsumsi lebih dari 200 kWh maka dikategorikan tinggi, jika berada di antara 100 hingga 200 kWh dikategorikan normal, dan jika kurang dari 100 kWh dikategorikan rendah.
- Variabel data_tidak_valid berfungsi sebagai penghitung jumlah data yang tidak valid selama proses pemeriksaan berlangsung.
- Fungsi print() digunakan untuk menampilkan hasil pemrosesan data ke layar, seperti nilai konsumsi listrik, status konsumsi, dan jumlah data yang tidak valid.
