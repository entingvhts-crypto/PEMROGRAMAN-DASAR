```python
# NAMA : ENTING VINCENCIA HUTASOIT
# NIM : 4232501038

# LATIHAN SOAL
# NO 1 (belanja buku)

jumlah_buku = int(input("masukkan jumlah buku: "))
harga_buku = int(input("masukkan harga per buku (Rp): "))
diskon_persen = int(input("masukkan diskon (%): "))
uang_budi = int(input("masukkan uang budi (Rp): "))
jumlah_orang = int(input("masukkan jumlah orang yang patungan: "))


total_belanja = jumlah_buku * harga_buku
besar_diskon = total_belanja * diskon_persen / 100
total_bayar = total_belanja - besar_diskon
uang_kembalian = uang_budi - total_bayar
iuran_per_orang = int(total_bayar) // jumlah_orang
sisa_uang = int(total_bayar) % jumlah_orang
uang_cukup = uang_budi >= total_bayar



print("\n===== HASIL PERHITUNGAN =====")
print("total belanja: Rp", int(total_belanja))
print("besar diskon: Rp", int(besar_diskon))
print("total yang harus dibayar: Rp", int(total_bayar))
print("uang budi: Rp", uang_budi)
print("uang kembalian: Rp", int(uang_kembalian))
print("iuran per orang: Rp", iuran_per_orang)
print("sisa pembagian: Rp", sisa_uang)
print("apakah uang budi cukup?", uang_cukup)

```

```python
#NO 2 (PLTS Atap Rumah)

jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_per_panel_wp = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan lama penyinaran efektif (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (contoh: 0.8): "))
kebutuhan_harian = float(input("Masukkan kebutuhan listrik per hari (kWh): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))

energi_harian_kwh = jumlah_panel * daya_per_panel_wp / 1000 * jam_penyinaran * efisiensi
energi_bulanan_kwh = energi_harian_kwh * 30
nilai_rupiah = energi_bulanan_kwh * tarif_listrik
kekurangan_harian = kebutuhan_listrik - energi_harian_kwh

cukup = energi_harian_kwh >= kebutuhan_listrik

print("\n===== HASIL PERHITUNGAN PLTS ====="1)
print("energi per hari (kwh):", round(energi_harian_kwh, 1))
print("energi per bulan (kwh):", round(energi_bulanan_kwh, 1))
print("nilai rupiah listrik yang dihasilkan: Rp", round(nilai_rupiah, 1))
print("kekurangan energi per hari (kwh):", round(kekurangan_harian, 1))
print("apakah PLTS mencukupi kebutuhan rumah?:", cukup)

```

```python
#NO 3 (Generator Diesel dan Rugi Kabel)

daya = float(input("masukkan daya generator (kw): "))
jam_operasi = int(input("masukkan lama operasi per hari (jam): "))
konsumsi_solar = float(input("masukkan konsusi solar (liter/kwh): "))
harga_solar = int(input("masukkan harga solar per liter (Rp): "))
kapasitas_tangki = int(input("masukkan kapasitas tangki (liter): "))

arus = float(input("masukkan arus kabel (A): "))
hambatan = float(input("masukkan hambatan kabel (ohm): "))

energi = daya * jam_operasi

solar_harian = energi * konsumsi_solar
biaya_harian = solar_harian * harga_solar

hari_penuh = int(kapasitas_tangki // solar_harian)
sisa_solar = kapasitas_tangki % solar_harian

rugi_daya = arus ** 2 * hambatan
rugi_tinggi = (rugi_daya > 150) and (arus >= 20)

print("\n===== HASIL PERHITUNGAN =====")
print(f"energi per hari : {energi:.2f} kwh")
print(f"kebutuhan solar per hari : {solar_harian:.2f} liter")
print(f"biaya solar per hari : Rp {biaya_harian:,.0f}")
print(f"tangki cukup untuk : {hari_penuh} hari penuh")
print(f"Sisa solar : {sisa_solar:.2f} liter")
print(f"Rugi daya kabel : {rugi_daya:.2f} W")
print(f"Rugi kabel tergolong tinggi: {rugi_tinggi}")

```
